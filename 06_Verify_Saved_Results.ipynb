This notebook verifies the completed checkpoint and displays recorded results without downloading model weights or running inference. Answer review remains AI-assisted, with two unresolved factual judgments and no independent human validation. Hash verification does not validate the correctness of those grades.

Run the cells in order. In Colab, the first cell clones the public project. Locally, start from the repository folder. The saved outputs below were produced by executing these same Python code cells during repository preparation, not by rerunning the original research models.


```python
# Get the repository in Colab, or use the local repository folder.
from pathlib import Path
import subprocess
import runpy

if Path('/content').exists():
    project = Path('/content/cybersecurity-rag-trust-scoring')
    if not project.exists():
        subprocess.run([
            'git', 'clone',
            'https://github.com/DarrellJ10/cybersecurity-rag-trust-scoring.git',
            str(project)
        ], check=True)
else:
    project = Path.cwd()

if not (project / 'verify_results.py').is_file():
    raise FileNotFoundError('Run this notebook from the repository folder.')
print('Repository files found. No models loaded.')
```

```python
# Verify all 430 checkpoint files and the saved prompt/answer records.
# This reads archives only; it does not extract or modify the research workspace.
verification = runpy.run_path(str(project / 'verify_results.py'), run_name='saved_results_verifier')
report = verification['verify'](project)
print('Integrity checks passed.')
print('Unique held-out model generations:', report['unique_generations'])
print('Policy abstention uses:', report['policy_abstention_uses'])
print('Total answer uses:', report['total_answer_uses'])
```

```python
# Display the original saved context metrics and AI-assisted answer grades.
# These summaries describe the archived run; they are not newly generated results.
verification['print_report'](report)
```
